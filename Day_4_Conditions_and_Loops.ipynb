# Day 4 - Conditions & Loops
**Topics:** `if`, `if-else`, `if-elif-else`, `match`, `while`, `for`, `break`, `continue`, `pass`, Patterns

## 1. if Statement

````python
age = 20

if age >= 18:
    print("Eligible to Vote")
````

````python
num = -5

if num < 0:
    print("Negative Number")
````

````python
marks = 82

if marks >= 35:
    print("Pass")
````

## 2. if-else

````python
num = 7

if num % 2 == 0:
    print("Even")
else:
    print("Odd")
````

````python
password = "python"

if password == "python":
    print("Login Successful")
else:
    print("Wrong Password")
````

````python
a = 25
b = 18

if a > b:
    print(a)
else:
    print(b)
````

## 3. if-elif-else

````python
marks = 86

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
else:
    print("Fail")
````

````python
signal = "yellow"

if signal == "red":
    print("Stop")
elif signal == "yellow":
    print("Ready")
elif signal == "green":
    print("Go")
else:
    print("Invalid Signal")
````

````python
day = 4

if day == 1:
    print("Monday")
elif day == 2:
    print("Tuesday")
elif day == 3:
    print("Wednesday")
elif day == 4:
    print("Thursday")
else:
    print("Other Day")
````

## 4. Nested if

````python
username = "admin"
password = "1234"

if username == "admin":
    if password == "1234":
        print("Login Successful")
    else:
        print("Wrong Password")
else:
    print("Unknown User")
````

````python
balance = 5000
amount = 2500

if balance >= amount:
    if amount > 0:
        print("Transaction Successful")
    else:
        print("Invalid Amount")
else:
    print("Insufficient Balance")
````

## 5. match-case

````python
day = 2

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case _:
        print("Invalid")
````

````python
operator = "*"

match operator:
    case "+":
        print(10 + 5)
    case "-":
        print(10 - 5)
    case "*":
        print(10 * 5)
    case "/":
        print(10 / 5)
    case _:
        print("Invalid Operator")
````

## 6. while Loop

````python
i = 1

while i <= 5:
    print(i)
    i += 1
````

````python
i = 5

while i >= 1:
    print(i)
    i -= 1
````

````python
i = 1
total = 0

while i <= 10:
    total += i
    i += 1

print(total)
````

````python
n = 7

i = 1
while i <= 10:
    print(f"{n} x {i} = {n*i}")
    i += 1
````

## 7. for Loop

````python
for i in range(5):
    print(i)
````

````python
for i in range(1, 6):
    print(i)
````

````python
for i in range(2, 11, 2):
    print(i)
````

````python
for ch in "Python":
    print(ch)
````

````python
fruits = ["Apple", "Mango", "Orange"]

for fruit in fruits:
    print(fruit)
````

## 8. Nested Loops

````python
for i in range(3):
    for j in range(3):
        print(i, j)
````

````python
for i in range(3):
    for j in range(3):
        print("*", end=" ")
    print()
````

## 9. break

````python
for i in range(10):
    if i == 5:
        break
    print(i)
````

````python
for ch in "Python":
    if ch == "h":
        break
    print(ch)
````

## 10. continue

````python
for i in range(1, 11):
    if i % 2 == 0:
        continue
    print(i)
````

````python
for ch in "Python":
    if ch == "o":
        continue
    print(ch)
````

## 11. pass

````python
if True:
    pass

print("Program Continues")
````

````python
for i in range(5):
    if i == 3:
        pass
    print(i)
````

## 12. Pattern 1 - Left Half Pyramid

````python
for i in range(5):
    for j in range(i + 1):
        print("*", end="")
    print()
````

## 13. Pattern 2 - Right Half Pyramid

````python
for i in range(5):
    for j in range(5 - i - 1):
        print(" ", end="")
    for j in range(i + 1):
        print("*", end="")
    print()
````

## 14. Pattern 3 - Full Pyramid

````python
n = 5

for i in range(n):
    for j in range(n - i - 1):
        print(" ", end="")
    for j in range(2 * i + 1):
        print("*", end="")
    print()
````
