# Data Types in Python

## What is a Data Type?

A **data type** defines:

* What kind of value is stored.
* How much memory it occupies.
* What operations can be performed on it.

Just like in real life:

* Age → Number
* Name → Text
* Is Student? → Yes/No

Similarly, Python stores different kinds of data using different data types.

---

## Why Do We Need Data Types?

Suppose we have:

```python
age = 22
```

Python understands that `22` is a number, so we can do:

```python
print(age + 10)
```

Output

```text
32
```

Now,

```python
name = "Satya"
```

If we write:

```python
print(name + 10)
```

Output

```text
TypeError
```

Because Python knows that `"Satya"` is a **string**, not an integer.

That is why data types are important.

## Arithmetic Operators

Used to perform mathematical operations.

```python
a = 10
b = 3

print("Addition       :", a + b)
print("Subtraction    :", a - b)
print("Multiplication :", a * b)
print("Division       :", a / b)
print("Modulus        :", a % b)
print("Floor Division :", a // b)
print("Exponent       :", a ** b)
```

**Output**

```text
Addition       : 13
Subtraction    : 7
Multiplication : 30
Division       : 3.3333333333333335
Modulus        : 1
Floor Division : 3
Exponent       : 1000
```

---

## Comparison Operators

Used to compare two values. The result is always `True` or `False`.

```python
a = 10
b = 20

print("a == b :", a == b)
print("a != b :", a != b)
print("a < b  :", a < b)
print("a > b  :", a > b)
print("a <= b :", a <= b)
print("a >= b :", a >= b)
```

---

## Logical Operators

Used to combine multiple conditions.

```python
x = True
y = False

print("x and y :", x and y)
print("x or y  :", x or y)
print("not x   :", not x)
```

---

## Assignment Operators

Used to assign or update values.

```python
a = 10

a += 5
print("+= :", a)

a -= 3
print("-= :", a)

a *= 2
print("*= :", a)
```

---

## Identity Operators

Used to check whether two variables refer to the same object in memory.

```python
a = [10, 20]
b = a
c = [10, 20]

print("a is b     :", a is b)
print("a is c     :", a is c)
print("a is not c :", a is not c)
```

This is the level I'd recommend for beginners: concise, readable, and each operator category gets its own explanation and example.

# Pattern Programs

## 1. Left Half Pyramid

```python
for i in range(5):
    for j in range(i + 1):
        print("*", end="")
    print()
```

**Output**

```text
*
**
***
****
*****
```

**Logic**

* Outer loop → Rows
* Inner loop → Prints `i + 1` stars

---

## 2. Inverted Left Half Pyramid

```python
for i in range(5):
    for j in range(i, 5):
        print("*", end="")
    print()
```

**Output**

```text
*****
****
***
**
*
```

**Logic**

* Outer loop → Rows
* Inner loop → Prints decreasing stars

---

## 3. Right Half Pyramid

```python
for i in range(5):
    for j in range(i, 5):
        print(" ", end="")
    for k in range(i + 1):
        print("*", end="")
    print()
```

**Output**

```text
     *
    **
   ***
  ****
 *****
```

**Logic**

* First loop → Prints spaces
* Second loop → Prints stars

---



**Logic**

* Spaces increase
* Stars decrease

---

## 5. Full Pyramid

```python
n = 5

for i in range(n):
    for j in range(n - i - 1):
        print(" ", end="")
    for j in range(2 * i + 1):
        print("*", end="")
    print()
```

**Output**

```text
    *
   ***
  *****
 *******
*********
```

**Logic**

* Spaces = `n - i - 1`
* Stars = `2*i + 1`
