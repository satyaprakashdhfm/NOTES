## 1. What is Data?

Data is a collection of facts or values. Examples include names, marks, phone numbers and addresses.

## What is a Data Structure?

A **Data Structure** is a way of storing and organizing data so it can be accessed and modified efficiently.

### Python Data Structures
- String
- List
- Tuple
- Set
- Dictionary

Today we will study **Strings**, one of the most commonly used data structures in programming and DSA.

## 2. What is a String?

A **String** is a sequence of characters enclosed inside single quotes (`' '`), double quotes (`" "`), or triple quotes.

### Characteristics
- Ordered
- Immutable
- Iterable
- Supports Indexing and Slicing
- Hashable (can be used as dictionary keys)


## 3. Creating Strings

````python
# Different ways to create strings

name = "Python"
city = 'Bengaluru'
message = '''Welcome to Python'''

# Convert another object into a string
number = 100

print(name)
print(city)
print(message)
print(str(number))
````

## 4. Indexing

Every character has a position called an **index**.

```
String :  P  y  t  h  o  n
Index  :  0  1  2  3  4  5
NegIdx : -6 -5 -4 -3 -2 -1
```

````python
text = "Python"

# Positive indexing
print(text[0])
print(text[3])

# Negative indexing
print(text[-1])
print(text[-3])
````

## 5. Slicing
Slicing extracts a part of a string.

````python
text = "Programming"

print(text[:])
print(text[:5])
print(text[3:8])
print(text[5:])
print(text[-5:])
print(text[:-2])
print(text[::2])
print(text[::-1])
````

## 6. Immutability

Strings are **immutable**, meaning their characters cannot be changed after creation.

Whenever you modify a string, Python creates a **new string** instead of changing the existing one.

````python
text = "Python"

# This is NOT allowed
# text[0] = "J"

# Create a new string instead
new_text = "J" + text[1:]

print(text)
print(new_text)
````

## 7. Hashable

A **hashable** object has a fixed hash value during its lifetime.

Since strings are immutable, they are hashable and can be used as dictionary keys.

````python
# Strings can be used as dictionary keys

student_marks = {
    "Alice": 90,
    "Bob": 85
}

print(student_marks["Alice"])
````

## 8. Looping Through Strings

````python
text = "Python"

# Using for loop # already told just a reminder
for ch in text:
    print(ch)
````

````python
text = "Code"

# Using while loop # already told just a reminder
i = 0
while i < len(text):
    print(text[i])
    i += 1
````

## 9. Membership Operators

````python
text = "Python"

# Check whether a substring exists
print("Py" in text)
print("Java" in text)
print("Java" not in text)
````

## 10. Useful Built-in Functions

````python
text = "Hello"

# Length of string
print(len(text))

# Concatenation
print(text + " World")

# Repetition
print(text * 3)
````

## 11. Escape Characters

````python
# New line
print("Hello\nWorld")

# Tab space
print("Name\tAge")

# Double quotes inside string
print("He said \"Python\"")
````

## 12. f-Strings

````python
name = "Alice"
marks = 95

# Insert variables directly into a string # already told just a reminder.
print(f"{name} scored {marks} marks.")
````

## Case Conversion
Short explanation and example.

````python
text="python programming"
# Convert to uppercase
print(text.upper())

# Convert to lowercase
print(text.lower())

# Capitalize first letter
print(text.capitalize())

# Convert every word to title case
print(text.title())
````

## Searching
Short explanation and example.

````python
text="Python Programming"

# Returns index or -1
print(text.find("Pro"))
print(text.find("Java"))

# Returns index, otherwise ValueError
print(text.index("Python"))
````

## Cleaning
Short explanation and example.

````python
text="   Python   "

# Remove spaces
print(text.strip())
print(text.lstrip())
print(text.rstrip())
````

## Replacing
Short explanation and example.

````python
text="I like Java"

# Replace old text with new text
print(text.replace("Java","Python"))
````

## Split and Join
Short explanation and example.

````python
text="apple,banana,mango"

# Convert string into list helpful in input taking in tcs exam to take comma separted inputs .
items=text.split(",")
print(items)

# Join list into string
print("-".join(items))
````

## Checking Methods
Short explanation and example.

````python
print("Python".isalpha())
print("12345".isdigit())
print("Python123".isalnum())
print("python".islower())
print("PYTHON".isupper())
print("   ".isspace())
print("Hello World".istitle())
````

## 13. Mini DSA Examples

````python
# Count vowels

text = "education"
count = 0

for ch in text.lower():
    if ch in "aeiou":
        count += 1

print("Vowels:", count)
````

````python
# Check palindrome

text = "madam"

if text == text[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
````

````python
# Count frequency of a character

text = "banana"

print(text.count("a"))
````

````python
# Count words

sentence = "Python is easy to learn"

print(len(sentence.split()))
````

## Summary

|---|---|
