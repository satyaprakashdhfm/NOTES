# **PART 1: MULTIMODAL AI**

## **1.1 Core Multimodal Representations**

### **Text Representation**
- **Traditional**: Bag-of-Words, TF-IDF (sparse vectors)
- **Embeddings**: Word2Vec, GloVe (dense semantic vectors)
- **Contextual**: BERT, GPT (dynamic context-aware)
- **Transformer encodings**: Self-attention for semantic relationships

### **Image Representation**
- **Classical features**: SIFT (Scale-Invariant Feature Transform), HOG (Histogram of Oriented Gradients)
- **Deep learning**: CNNs extract hierarchical features (edges → textures → objects)
- **Pre-trained models**: ResNet, EfficientNet, Vision Transformers (ViT)
- **Cross-modal**: CLIP maps images + text into shared embedding space

### **Audio Representation**
- **Waveforms**: Raw temporal signals
- **Spectrograms**: Time-frequency visual representations
- **MFCC**: Mel-Frequency Cepstral Coefficients (acoustic features)
- **Deep learning**: End-to-end from raw audio (Wav2Vec)

### **Video Representation**
- **Frame-based**: Process individual frames as images
- **Temporal**: Motion features, optical flow
- **3D CNNs**: Spatiotemporal convolutions
- **Two-stream networks**: Separate spatial + temporal pathways

---

## **1.2 Fusion Strategies**

### **Early Fusion (Feature-level)**
- Concatenate features from all modalities before training
- **Pros**: Captures low-level cross-modal interactions
- **Cons**: Scale mismatches, computational complexity
- **Use**: Video classification, sentiment analysis

### **Late Fusion (Decision-level)**
- Train separate models per modality → Combine predictions
- **Key characteristic**: Independent modality-specific models
- **Pros**: Flexibility, handles missing modalities
- **Cons**: May miss fine-grained interactions
- **Use**: Sensor fusion, ensemble systems

### **Hybrid Fusion**
- Multi-stage integration (early + late)
- Balances interaction and independence

### **Cross-Modal Attention**
- One modality attends to specific regions of another
- Text queries relevant image regions (Visual Question Answering)
- **Mechanism**: Query-Key-Value framework

### **Tensor Fusion**
- High-dimensional tensor products capture complex interactions
- Expensive but expressive (unimodal, bimodal, trimodal)

---

## **1.3 Deep Learning Architectures**

### **CNNs** - Extract hierarchical spatial features from images
### **RNNs** - Process sequences (LSTM/GRU for temporal dependencies)
### **Transformers** - Self-attention, parallel processing, long-range dependencies
### **Joint Embeddings** - Shared representation space across modalities
### **Attention Mechanisms** - Selective focus (self-attention, cross-attention)

## **2.1 What is Agent Persona?**

**Definition**: Structured definition of agent's identity, behavior, and boundaries

**Purpose**: Ensure consistent, predictable, safe agent responses

---

## **2.2 Why Critical?**

- **Contextual Control**: Defines operational scope
- **Consistency** ✓: Valid performance indicator, uniform responses

- **User Experience**: Predictable, professional interactions

- **Policy Compliance**: Ethical guidelines, regulations

---

## **2.3 Blueprint Components**

### **Identity**
- **Name & Designation**: Defines role and authority
- **Primary purpose**: Define identity and role in system

### **Specialization** ✓
- Domain expertise, strengths, reasoning style
- **Where to define**: Strengths & reasoning style go here

### **Execution**
- Capabilities, decision authority, tool access

- **Where to define**: Permissions like "read user_data" go here

### **Interaction**

### **Boundaries**
- **Knowledge**: In-scope vs out-of-scope

- **Bias awareness**: Known biases, mitigation strategies

# **PART 3: AGENTIC DESIGN**

## **3.1 Types of Agents**

### **Reactive Agents**
- Respond to stimuli without future modeling
- Immediate action, no internal state
- Fast, simple, rule-based

### **Planning-Based Agents**
- Generate sequence of steps before acting
- Create structured roadmap, goal-oriented
- Complex tasks requiring foresight

### **Memory-Based Agents**
- Store past interactions for personalization
- Stateful, contextual awareness
- Long-term relationships

### **Reflection-Based Agents**
- Self-critique and iterative improvement
- Meta-cognition, quality assurance
- High-accuracy tasks

### **Tool-Using Agents**
- Leverage external APIs/tools
- Action-oriented, capability extension
- Specialized tasks

### **Conditional Agents**
- If-then rules, deterministic branching
- Workflow automation

### **Multi-Agent Systems**
- Multiple specialists collaborate
- Delegation, coordination

---

## **3.2 Design Patterns**

### **ReAct (Reason + Act)**
- **Process**: Thought → Action → Observation (loop)
- **Focus**: Adaptive iteration
- **When**: Dynamic environments, trial-and-error
- **Example**: Search query → Read result → Refine search

### **Planning Pattern**
- **Process**: Goal → Plan all steps → Execute
- **Focus**: Upfront strategy, structured roadmap
- **When**: Multi-step coordination, dependencies
- **Difference from ReAct**: Plans everything first vs iterative

### **Memory Pattern**
- **Components**: Short-term (conversation), Long-term (preferences)
- **Purpose**: Store past conversations for personalization
- **Focus**: Context retention
- **When**: Personalized assistants, long sessions

### **Reflection Pattern**
- **Process**: Generate → Self-Critique → Revise → Repeat
- **Focus**: Quality improvement, self-awareness
- **When**: High-stakes tasks, accuracy required

### **Tool Usage Pattern**
- **Process**: Assess → Select tool → Execute → Integrate
- **Focus**: Capability extension
- **When**: Tasks beyond LLM native abilities
- **Examples**: Web search, calculator, database

### **Delegation Pattern**
- **Process**: Task decomposition → Route to specialists → Aggregate
- **Role**: Assign subtasks to specialized agents
- **When**: Complex tasks benefit from experts

### **Multi-Agent Pattern**
- Multiple agents with distinct roles collaborate
- **When useful**: Diverse expertise, parallel processing, complex workflows
- **Frameworks**: CrewAI, AutoGen, LangGraph

### **Observability Pattern**
- **Purpose**: Log system performance for debugging
- **Tracks**: Agent thoughts, tool calls, errors, latency, costs
- **When**: Production systems requiring transparency

---

## **3.3 Key Agent Features**

**Tools**: External functions/APIs (search, calculator, database)  
**Memory**: Context retention (short-term + long-term)  
**Reasoning**: Logical inference, chain-of-thought  
**Decision-Making**: Action selection based on goals

# **PART 1: SOFTWARE DEVELOPMENT LIFECYCLE MODELS**

## **1.1 Waterfall Model**
- **Sequential phases**: Requirements → Design → Implementation → Testing → Deployment → Maintenance
- **Characteristics**: Linear, rigid, documentation-heavy
- **Pros**: Clear structure, well-documented
- **Cons**: Inflexible to changes, late testing, high risk
- **Best for**: Stable requirements, regulated industries

## **1.2 Agile Model**
- **Iterative & incremental**: Work in short cycles (sprints)
- **Characteristics**: Adaptive, collaborative, customer-focused
- **Pros**: Flexible to changes, early delivery, continuous feedback
- **Cons**: Less documentation, requires discipline
- **Best for**: Evolving requirements, fast-paced projects

## **1.3 Spiral Model**
- **Risk-driven approach**: Plan → Risk Analysis → Engineering → Evaluation (repeat)
- Combines iterative + waterfall elements
- **Best for**: High-risk, large-scale projects

## **1.4 Prototyping Model**
- Build quick prototype → Gather feedback → Refine
- **Best for**: Unclear requirements, UI-heavy applications

# **PART 6: REFACTORING**

## **6.1 What is Refactoring?**

**Definition**: Improving code structure without changing external behavior

**Goal**: Enhance readability, reduce complexity, eliminate duplication

---

## **6.2 When We Need Refactoring**

- Code smells detected (duplicates, long methods, complex conditionals)
- Before adding new features (clean foundation)
- After getting tests to pass (improve design)
- During code reviews
- When maintenance becomes difficult

---

## **6.3 Why We Need Refactoring**

- **Improve maintainability**: Easier to understand and modify
- **Reduce technical debt**: Pay down accumulated design shortcuts
- **Enable future changes**: Flexible codebase adapts better
- **Enhance performance**: Optimize bottlenecks
- **Increase code quality**: Cleaner, more elegant solutions

---

## **6.4 Technical Debt**

**Definition**: Cost of choosing quick/easy solution over better long-term approach

**Causes**:
- Rushed deadlines
- Lack of refactoring
- Poor design decisions
- Accumulated code smells

**Consequences**:
- Slower development over time
- More bugs
- Higher maintenance costs
- Team frustration

**Management**: Regular refactoring pays down technical debt incrementally

---

## **6.5 How to Refactor in Agile**

1. **Ensure tests exist** - Safety net before changes
2. **Small incremental steps** - One refactoring at a time
3. **Run tests after each change** - Verify behavior unchanged
4. **Commit frequently** - Version control checkpoints
5. **Refactor during sprint** - Continuous improvement, not separate phase

---

## **6.6 Refactoring Techniques**

### **Extract Method**
**Problem**: Long method doing multiple things

**Before**:
```java
void printOwing() {
    printBanner();
    // Print details
    System.out.println("Name: " + name);
    System.out.println("Amount: " + getOutstanding());
}
```

**After**:
```java
void printOwing() {
    printBanner();
    printDetails(getOutstanding());
}

void printDetails(double outstanding) {
    System.out.println("Name: " + name);
    System.out.println("Amount: " + outstanding);
}
```

---

### **Extract Class**
**Problem**: One class doing work of two

**Before**:
```java
class Person {
    String name;
    String officeAreaCode;
    String officeNumber;
    
    String getTelephoneNumber() {
        return "(" + officeAreaCode + ") " + officeNumber;
    }
}
```

**After**:
```java
class Person {
    String name;
    TelephoneNumber officeTelephone;
}

class TelephoneNumber {
    String areaCode;
    String number;
    
    String getTelephoneNumber() {
        return "(" + areaCode + ") " + number;
    }
}
```

---

### **Inline Class**
**Problem**: Class doing too little, not worth keeping separate

**Before**:
```java
class Person {
    TelephoneNumber phone;
}
class TelephoneNumber {
    String number;
}
```

**After**:
```java
class Person {
    String phoneNumber;
}
```

---

### **Self-Encapsulate Field**
**Problem**: Direct field access, hard to add logic later

**Before**:
```java
class Rectangle {
    int width;
    int height;
    
    int area() {
        return width * height; // Direct field access
    }
}
```

**After**:
```java
class Rectangle {
    private int width;
    private int height;
    
    int getWidth() { return width; }
    void setWidth(int w) { width = w; }
    
    int area() {
        return getWidth() * getHeight(); // Use getters
    }
}
```

**Image 2 Answer**: **Self-Encapsulate field** - The static analyzer flags direct field access in lines 4 and 5 ✓

---

### **Replace Array with Object**
**Problem**: Array elements represent different concepts

**Before**:
```java
String[] row = new String[3];
row[0] = "Liverpool";
row[1] = "15";
```

**After**:
```java
class Performance {
    String name;
    int wins;
}
Performance row = new Performance();
row.name = "Liverpool";
row.wins = 15;
```

---

### **Replace Conditional with Polymorphism**
**Problem**: Complex if-else or switch based on type

**Before**:
```java
double getSpeed() {
    switch (type) {
        case EUROPEAN: return getBaseSpeed();
        case AFRICAN: return getBaseSpeed() - getLoadFactor();
        case NORWEGIAN: return (isNailed) ? 0 : getBaseSpeed();
    }
}
```

**After**:
```java
abstract class Bird {
    abstract double getSpeed();
}
class European extends Bird {
    double getSpeed() { return getBaseSpeed(); }
}
class African extends Bird {
    double getSpeed() { return getBaseSpeed() - getLoadFactor(); }
}
```

---

### **Move Method**
**Problem**: Method used more by another class than its own

**Before**:
```java
class Account {
    double overdraftCharge() {
        return type.overdraftCharge(daysOverdrawn); // Uses type more
    }
}
```

**After**:
```java
class AccountType {
    double overdraftCharge(int daysOverdrawn) {
        // Method moved here
    }
}
class Account {
    double overdraftCharge() {
        return type.overdraftCharge(daysOverdrawn);
    }
}
```

# **PART 7: TEST-DRIVEN DEVELOPMENT (TDD)**

## **7.1 What is TDD?**

**Definition**: Write tests before writing production code

**Mantra**: Red → Green → Refactor

---

## **7.2 TDD Cycle (Characteristic Sequence)**

1. **Write a failing test** (Red) ✓
   - Define expected behavior
   - Test fails because code doesn't exist yet

2. **Write minimal code to pass** (Green) ✓
   - Make test pass with simplest implementation
   - Focus on functionality, not elegance

3. **Refactor** (Blue) ✓
   - Improve code structure
   - Maintain passing tests
   - Pay down technical debt

4. **Repeat** - Next feature/test

---

## **7.3 Features of TDD**

- **Test-first approach**: Tests drive design
- **Incremental development**: Small steps
- **Continuous validation**: Always have working code
- **Living documentation**: Tests describe behavior
- **Design emerges**: Architecture evolves from requirements

---

## **7.4 Advantages**

- **Fewer bugs**: Catch issues immediately
- **Better design**: Forces modular, testable code
- **Confidence**: Safe to refactor
- **Faster debugging**: Precise failure location
- **Regression prevention**: Tests catch breaking changes

---

## **7.5 Shortcomings**

- **Initial slowdown**: Takes time to write tests first
- **Learning curve**: Requires discipline and practice
- **Maintenance overhead**: Tests need updates too
- **Not suitable for all**: Exploratory coding, prototypes
- **False confidence**: Tests can have bugs too

---

## **7.6 TDD vs Traditional Testing**

| **Aspect** | **Traditional Testing** | **TDD** |
|------------|-------------------------|---------|
| **When** | After code is written | Before code exists ✓ |
| **Focus** | Finding bugs | Preventing bugs, driving design ✓ |
| **Design** | Code-first, test later | Test-first, emergent design ✓ |
| **Coverage** | Often incomplete | High coverage by default |
| **Refactoring** | Risky without tests | Safe with test safety net |
| **Development** | Speculative programming risk ✓ | Focused, incremental |

**Key Difference**: TDD ensures **speculative programming is avoided** - only write code needed to pass tests ✓

## **7.8 Mocking in TDD**

### **What is Mocking?**
**Definition**: Creating fake objects that simulate behavior of real dependencies

### **Why Mock?**
- **Isolate unit under test**: Remove external dependencies
- **Control behavior**: Simulate specific scenarios (errors, edge cases)
- **Speed**: Avoid slow operations (database, network)
- **Focus**: Test one component at a time

### **Example**:
```java
// Without mocking - tests database (slow, fragile)
@Test
public void testGetUser() {
    UserService service = new UserService();
    User user = service.getUser(1); // Hits real database
    assertEquals("John", user.getName());
}

// With mocking - tests logic only (fast, reliable)
@Test
public void testGetUser() {
    UserRepository mockRepo = mock(UserRepository.class);
    when(mockRepo.findById(1)).thenReturn(new User("John"));
    
    UserService service = new UserService(mockRepo);
    User user = service.getUser(1);
    assertEquals("John", user.getName());
}
```

# **PART 8: SOLID PRINCIPLES**

## **8.1 What are SOLID Principles?**

**Definition**: Five design principles for writing maintainable, scalable object-oriented code

**Purpose**: Guide software design to reduce complexity and improve flexibility

---

## **8.2 SOLID Breakdown**

### **S - Single Responsibility Principle (SRP)** ✓

**Definition**: A class should have only one reason to change

**Explanation**: Each class does one thing well

**Problem**: MyMail class handles email details AND multiple protocols (POP3, IMAP, etc.)

**Violation Example**:
```java
class MyMail {
    // Email properties
    private String subject;
    private String body;
    
    // Multiple responsibilities - VIOLATION
    void sendViaPOP3() { /* ... */ }
    void sendViaIMAP() { /* ... */ }
    void sendViaSMTP() { /* ... */ }
}
```

**Solution**:
```java
class Email {
    private String subject;
    private String body;
    // Only manages email data
}

class EmailSender {
    void send(Email email, Protocol protocol) {
        // Delegates to specific protocol handlers
    }
}

interface Protocol {
    void send(Email email);
}
class POP3Protocol implements Protocol { /* ... */ }
class IMAPProtocol implements Protocol { /* ... */ }
```

**Image 1 Answer**: **Single Responsibility Principle** is violated ✓ - MyMail does too much

---

### **O - Open/Closed Principle (OCP)** ✓

**Definition**: Open for extension, closed for modification

**Explanation**: Add new functionality without changing existing code

**Example**:
```java
// BAD - Must modify class to add new shape
class AreaCalculator {
    double calculateArea(Object shape) {
        if (shape instanceof Circle) {
            // Circle area logic
        } else if (shape instanceof Rectangle) {
            // Rectangle area logic
        }
        // Must modify this method for new shapes!
    }
}

// GOOD - Extend without modifying
interface Shape {
    double area();
}

class Circle implements Shape {
    public double area() { return Math.PI * radius * radius; }
}

class Rectangle implements Shape {
    public double area() { return width * height; }
}

// No modification needed for new shapes!
class Triangle implements Shape {
    public double area() { return 0.5 * base * height; }
}
```

---

### **L - Liskov Substitution Principle (LSP)** ✓

**Definition**: Subtypes must be substitutable for their base types

**Explanation**: Child classes shouldn't break parent class contracts

**Example**:
```java
// VIOLATION
class Bird {
    void fly() { /* flying logic */ }
}

class Penguin extends Bird {
    @Override
    void fly() {
        throw new Exception("Can't fly!"); // Breaks contract!
    }
}

// CORRECT
interface Bird {
    void eat();
}

interface FlyingBird extends Bird {
    void fly();
}

class Sparrow implements FlyingBird {
    public void fly() { /* fly */ }
}

class Penguin implements Bird {
    public void eat() { /* eat */ }
    // No fly method - doesn't violate expectations
}
```

---

### **I - Interface Segregation Principle (ISP)** ✓

**Definition**: Clients shouldn't depend on interfaces they don't use

**Explanation**: Many specific interfaces > One general interface

**Example**:
```java
// BAD - Fat interface
interface Worker {
    void work();
    void eat();
    void sleep();
}

class Robot implements Worker {
    public void work() { /* work */ }
    public void eat() { /* Robots don't eat! */ }
    public void sleep() { /* Robots don't sleep! */ }
}

// GOOD - Segregated interfaces
interface Workable {
    void work();
}

interface Eatable {
    void eat();
}

class Human implements Workable, Eatable {
    public void work() { /* work */ }
    public void eat() { /* eat */ }
}

class Robot implements Workable {
    public void work() { /* work */ }
    // Only implements what it needs
}
```

**Image 1 Context**: If MyMail forced clients to implement all protocols even when they need just one, ISP would be violated

---

### **D - Dependency Inversion Principle (DIP)** ✓

**Definition**: Depend on abstractions, not concretions

**Explanation**: High-level modules shouldn't depend on low-level details

**Example**:
```java
// BAD - Depends on concrete class
class EmailService {
    private MySQLDatabase database; // Concrete dependency
    
    void saveEmail(Email email) {
        database.save(email);
    }
}

// GOOD - Depends on abstraction
class EmailService {
    private Database database; // Abstract dependency
    
    EmailService(Database db) {
        this.database = db;
    }
    
    void saveEmail(Email email) {
        database.save(email);
    }
}

interface Database {
    void save(Email email);
}

class MySQLDatabase implements Database { /* ... */ }
class MongoDatabase implements Database { /* ... */ }
// Easy to swap implementations
```

---

## **8.3 SOLID Summary**

| **Principle** | **Key Idea** | **Benefit** |
|---------------|--------------|-------------|
| **S**RP | One class, one job | Easy to understand, modify |
| **O**CP | Extend, don't modify | Add features without risk |
| **L**SP | Subclasses honor contracts | Safe polymorphism |
| **I**SP | Small, focused interfaces | No unused dependencies |
| **D**IP | Depend on abstractions | Flexible, testable code |

# **PART 9: DESIGN PATTERNS**

## **9.1 Singleton Pattern**

**Definition**: Ensure a class has only one instance with global access point

**Use case**: Database connections, logging, configuration

**Example**:
```java
class Singleton {
    private static Singleton instance;
    
    private Singleton() {} // Private constructor
    
    public static Singleton getInstance() {
        if (instance == null) {
            instance = new Singleton();
        }
        return instance;
    }
}
```

---

## **9.2 Factory Pattern**

**Definition**: Create objects without specifying exact class

**Use case**: Object creation logic varies based on conditions

**Example**:
```java
interface Shape {
    void draw();
}

class Circle implements Shape {
    public void draw() { System.out.println("Circle"); }
}

class ShapeFactory {
    public Shape getShape(String type) {
        if (type.equals("CIRCLE")) return new Circle();
        // Add more shapes without changing client code
        return null;
    }
}
```
